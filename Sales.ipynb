# create new business metrices

```python
df['Sales Revenue']=(df['Units Sold']*df['Price'])
```

```python
df['Price Difference']=(df['Price'] - df['Competitor Pricing'])
```

```python
df['Inventory to sales ratio']=np.where(df['Units Sold']>0,df['Inventory Level']/df['Units Sold'],np.nan)
```

# Exploratory Data Analysis (EDA)

# category wise sales performances

```python
category_sales=df.groupby('Category').agg(
    Total_Units_Sold=('Units Sold','sum'),
    Total_Revenue=('Sales Revenue','sum'),
    Average_Price=('Price','mean')
).reset_index()
```

```python
category_sales
```

```python
category_sales=category_sales.sort_values(by='Total_Revenue',ascending=False)
```

```python
print(category_sales)
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=category_sales,x='Category',y='Total_Revenue')

plt.title("Revenue by product category")
plt.xticks(rotation=45)
plt.show()
```

# Region wise performance

```python
region_sales=df.groupby('Region').agg(
    Total_Units=('Units Sold','sum'),
    Total_Revenue=('Sales Revenue','sum')
).reset_index()
```

```python
region_sales
```

# Monthly sales trends

```python
monthly_sales=df.groupby(
    ['Year','Month']
)['Sales Revenue'].sum().reset_index()

```

```python
monthly_sales
```

```python
monthly_sales['Period']=pd.to_datetime(
    monthly_sales['Year'].astype(str) + '-'+
    monthly_sales['Month'].astype(str) + '-01'
)
```

```python
monthly_sales=monthly_sales.sort_values('Period')

plt.figure(figsize=(14,5))
sns.lineplot(data=monthly_sales,x='Period',y='Sales Revenue',markers='o')

plt.title("monthly sales revenue trend")
plt.xticks(rotation=45)
plt.show()
```

# promotion impact

```python
promotion_analysis = df.groupby('Promotion').agg(
    Average_Units_Sold=('Units Sold', 'mean'),
    Average_Revenue=('Sales Revenue', 'mean'),
    Total_Units_Sold=('Units Sold', 'sum')
).reset_index()

print(promotion_analysis)
```

# Inventory analysis

```python
inventory_analysis = df.groupby('Category').agg(
    Average_Inventory=('Inventory Level', 'mean'),
    Average_Units_Sold=('Units Sold', 'mean'),
    Total_Demand=('Demand', 'sum')
).reset_index()

print(inventory_analysis)
```
